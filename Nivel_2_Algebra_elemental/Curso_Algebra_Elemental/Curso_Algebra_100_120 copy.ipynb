## Video 109
### Cómo se resuelve una Ecuación con x en el exponente y con fracciones
$$
\left( \frac{2}{3}\right)^{x+6} = \left( \frac{9}{4}\right)^{\frac{2x+1}{2}} \left( \frac{8}{27}\right)^{\frac{1+x}{3}}
$$



```python
expr_raw = r" \left( \frac{2}{3}\right)^{x+6} = \left( \frac{9}{4}\right)^{\frac{2x+1}{2}} \left( \frac{8}{27}\right)^{\frac{1+x}{3}} "
LHS = sp.Rational(2, 3) ** (x + 6)
RHS = sp.Rational(9, 4) ** ((2 * x + 1) / 2) * sp.Rational(8, 27) ** ((1 + x) / 3)
eq = sp.Eq(LHS, RHS)
solution = sp.solve(eq, x)[0]

# result = (pol1 * pol2).expand()
display(Math(rf"{expr_raw} {spacing_implies} x = {sp.latex(solution)}"))
```

$$
\left( a^6 b^{4/7} c^{2/3}\right)^\frac{9}{4}
$$



```python
a, b, c = sp.symbols("a, b, c", positive=True)
expr_raw = r" \left( a^6 b^{4/7} c^{2/3}\right)^\frac{9}{4}"
expr = (a**6 * b ** sp.Rational(4, 7) * c ** sp.Rational(2, 3)) ** sp.Rational(9, 4)

display(Math(rf"{expr_raw} {spacing_equal} {sp.latex(expr)}"))
```
