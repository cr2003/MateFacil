## Video 122
###  ¿Qué son los Productos Notables?

Son multiplicaciones de polinomios, que surgen con mucha frecuentia y que pueden resolverse rápidamente con reglas sencillas.


#### Lista de productos notables:

* Binomios conjugados $(a+b)(a-b) $

* Binomio al cuadrado $(a+b)^2$

* Binomio con término común $(a+b)(a+c)$

* Binomio al cubo $(a+b)^3$

* Binomio a la n (Binomio de Newton) $(a+b)^n$

* Trinomio al cuadrado $(a+b+c)^2$

* Polinomio al cuadrado $(a_1 + a_2 + \cdots + a_m)^2$

* Polinomio a la n $(a_1 + a_2 + \cdots + a_m)^n$

## Video 129
### Binomio de Newton: Calcular el coeficiente $x^{210}$

El binonio de Newton:

$$ (a+b)^n = \binom{n}{0}a^n + \binom{n}{1}a^{n-1}b+ \binom{n}{2}a^{n-2}b^2 + \cdots + \binom{n}{k}a^{n-k}b^k +
\cdots + \binom{n}{n}b^n $$

$$ (a-b)^n = \binom{n}{0}a^n - \binom{n}{1}a^{n-1}b+ \binom{n}{2}a^{n-2}b^2 - \cdots + (-1)^k\binom{n}{k}a^{n-k}b^k +
\cdots + (-1)^n\binom{n}{n}b^n $$

Coeficiente binomial:
$$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$











a) Calcular el coeficiente de $x^{210}$ en el desarrollo de $(x-x^3)^{90}$

```python
# Define the expression
expr = (x - x**3) ** 90

# Method 1: Expand and extract the coefficient directly
# SymPy's .coeff(variable, power) extracts the coefficient of a specific term
coefficient_method1 = expr.expand().coeff(x, 210)

print(f"Method 1 (Extraction): {coefficient_method1}")
```

```python
# Method 2: Compute it analytically using the Binomial Theorem formula
# The general term is: (-1)**k * binomial(n, k) * x**(90 - k) * (x**3)**k

k, x = sp.symbols("k, x", integer=True)

# From the problem statement, de components of the general term: a^(n-k) * b^k
# where a=x, b=x^3, and n=90
general_term_x = (x ** (90 - k)) * ((x**3) ** k)

# Combine the powers of x into a single exponent (converts to x**(2*k + 90))
combined_term = sp.powsimp(general_term_x)

# Extract the base (x) and the exponent (2*k + 90)
base, exponent = combined_term.as_base_exp()

# Solve for when the exponent equals 210
k_value = sp.solve(sp.Eq(exponent, 210), k)[0]

print(f"The automated value of k is: {k_value}")
```

```python
# Method 2: Compute it analytically using the Binomial Theorem formula
# The general term is: (-1)**k * binomial(n, k) * x**(90 - k) * (x**3)**k
# As calculated previously, n = 90 and k = 60
n = 90
k = 60

coefficient_method2 = (-1) ** k * sp.binomial(n, k)

print(f"Method 2 (Formula):    {coefficient_method2}")

# Verify both methods yield the exact same large integer
assert coefficient_method1 == coefficient_method2
```

## Video 130
### Binomio de Newton: Calcular el término que no tiene x


b) Calcular el término que no contiene x en el desarrollo de:

$$\left( \sqrt{x} + \frac{1}{\sqrt[4]{x}} \right)^9$$

```python
# Method 1: Expand and extract the coefficient directly

# Define the symbol (positive=True helps with fractional power simplifications)
x = sp.symbols("x", positive=True)

# Define the expression: (sqrt(x) + 1 / x**(1/4))**9
expr = (sp.sqrt(x) + 1 / (x ** (1 / 4))) ** 9

# Expand and find the coefficient of x**0
constant_term = expr.expand().coeff(x, 0)

print(f"The term independent of x is: {constant_term}")
```

```python
# Method 2: Compute it analytically using the Binomial Theorem formula

# Define symbols
x, k = sp.symbols("x k", positive=True)

# Write the term's x components: (x**(1/2))**(9-k) * (x**(-1/4))**k
general_term_x = (x ** (sp.Rational(1, 2))) ** (9 - k) * (
    x ** (sp.Rational(-1, 4))
) ** k

# Combine powers to isolate the total exponent
combined = sp.powsimp(general_term_x)
_, exponent = combined.as_base_exp()

# Solve exponent = 0 to find k
k_value = sp.solve(sp.Eq(exponent, 0), k)[0]

# Compute the binomial coefficient: binomial(9, k)
coefficient = sp.binomial(9, k_value)

print(f"Calculated k: {k_value}")
print(f"The coefficient is: {coefficient}")
```
